```python
import random

print("Pensei em um número entre 1 e 5. Tente adivinhar!")

# O computador escolhe o número
num_secreto = random.randint(1, 5)

# O jogador tem 3 chances
for tentativa in range(1, 4):
    print(f"\nTentativa {tentativa} de 3")

    # Recebemos o chute e transformamos em número inteiro (int)
    chute = int(input("Digite seu número: "))

    if chute == num_secreto:
        print("🎉 Parabéns! Você acertou o número!!")
        break  # Interrompe o loop se a pessoa acertar
    else:
        if tentativa < 3:
            print("Errou! Tente de novo.")
        else:
            print(f"Que pena! Suas chances acabaram. O número era {num_secreto}.")
```

```python
# Jodo da IA - Resolução
# concurso dataprev

import random

numero_ia = random.randint(1, 5)

print("--- JOGO DA IA ----")
print("Eu pensei em um número de 1 a 5. Você tem 3 chances para adivinhar!")

for tentativa in range(3):
    palpite = int(input("Qual é o seu palpite? "))

    if palpite == numero_ia:
        print("UAU! Você acertou!")
        break # Quebra o laço de repetição se acertar
    else:
        print("Errou! Tente novamente.")

print("O número secreto era:", numero_ia)
```
