```python
# Dados para a dinâmica
import pandas as pd

# Histórico dos meses anteriores
dados_marketing = {
    'Investimento_Anuncios': [1000, 2000, 3000, 4000, 5000],
    'Faturamento': [15000, 25000, 36000, 48000, 58000]
}

tabela_mkt = pd.DataFrame(dados_marketing)
display(tabela_mkt)

```

```python
# Gráfico de dispersão
tabela_mkt.plot(kind='scatter', x='Investimento_Anuncios', y='Faturamento', color='orange', title='Investimento vs Faturamento')
```

```python
# Preparação e Treino
from sklearn.linear_model import LinearRegression

X = tabela_mkt[['Investimento_Anuncios']]
y = tabela_mkt['Faturamento']

modelo_mkt = LinearRegression()
modelo_mkt.fit(X, y)

investimento_futuro= pd.DataFrame({'Investimento_Anuncios':[8000]})
predicao = modelo_mkt.predict(investimento_futuro)

print(f"A IA prevê que o Faturamento vai ser de {predicao}")
```
