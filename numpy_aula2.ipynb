```python
#  A matemática da IA (Numpy)

import numpy as np

print("--- O PROBLEMA DO PYTHOM PUROA ---")
precos_lista = [100000, 200000, 300000]

# O Python puro não faz a conta, ele apenas duplica os elementos de lista!
print("Lista vezes 2: ", precos_lista * 2)

print("\n--- A SOLUÇÃO COM NUMPY ---")
# Trasformas a lista em um 'Array" (um super matemático)
precos_numpy = np.array([100000, 200000, 300000])

# O Numpy percebe instataneamente que  queremos multiplicar cada número!
precos_atualizados = precos_numpy * 2
print("Array vezes 2:", precos_atualizados)

print("\n--- COMO A IA VÊ O MUNDO (MATRIZES) ---")
# Na IA (especialmente em imagens), tudo são números (Matrizes)
matriz_dados = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(matriz_dados)
```
