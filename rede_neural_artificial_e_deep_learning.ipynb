```python
# 1. Carregando os Dados do Banco
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Lendo o arquivo CVS
tabela_banco = pd.read_csv('historico_credito_100_clientes.csv')

print("Visualizando os primeiros 5 clientes da base de dados:\n")
display(tabela_banco.head())

# 2. Separando X (Características) e y (Alvo)
X = tabela_banco[['Idade', 'Salario', 'Divida_Atual']]
y = tabela_banco['Risco_Alto']

# 3. Normalizando os dados (OBRIGATÓRIO para Redes Neurais)
normalizador = StandardScaler()
X_normalizando = normalizador.fit_transform(X)

# 4. Separando Treino e Teste (Regra de Ouro)
X_treino, X_teste, y_treino, y_teste = train_test_split(X_normalizando, y, test_size=0.2, random_state=42)

print("\nBase com 100 clientes pronta e normalizada para a Rede Neural!")
```

```python
# 2. Contruindo a Arquitetura de Rede Neural Artificial (O Cérebro)
import tensorflow as tf
from tensorflow import keras

# 1. Inicializando a Rede neural Sequencial (camadas em sequência)
modelo_rn = keras.Sequential([
    # Temos 3 informações de entrada: Idade, Salario e Dívida Atual
    keras.Input(shape=(3,)),

    # 1ª Camada Oculta (Quantidade de neurônios)
    keras.layers.Dense(6, activation='relu'),

    # Camada de Saida (O nosso espremedor de probabilidade: 0 a 1)
    keras.layers.Dense(1, activation='sigmoid')
])

# 2. Compilando o cérebro (Definindo as regras de estudo0
modelo_rn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

print("Cérebro digital construido com sucesso! Veja o resumo abaixo!")
modelo_rn.summary()
```

```python
# 3. O Treinamento (Deixando a máquina estudar

# Passamos os dados do treino. Vamos pedir para ela ler tudo 50 vezes (epochs-50)
print("Iniciando o treinamento da Rede neural ...\n")
historico = modelo_rn.fit(X_treino, y_treino, epochs=50)

print("\nTreinamento concluido!")

# Fazendo a prova final com os daos de teste escondidos
perda, acuracia = modelo_rn.evaluate(X_teste, y_teste)
print(f"\nAcurácia na Prova Final: {acuracia * 100:.2f}% de acerto")
```

```python
# 4. Simulando a Análise de um Novo Cliente

# 1. Entrada de dados de um cliente solicitando crédito hoje
novo_cliente = {
    'Idade': [26],
    'Salario': [3000],
    'Divida_Atual': [1000]
}

novo_cliente_df = pd.DataFrame(novo_cliente)

# 2. Passo essencial: Passar o cliente pelo mesmo padronizador matemático
novo_cliente_normalizado = normalizador.transform(novo_cliente_df)

# 3. Pedondo para a Rede neural calcular a probabilidade de risco
predicao = modelo_rn.predict(novo_cliente_normalizado)
probabilidade_risco = predicao[0][0]

print(f"Relatorio de Análise: {probabilidade_risco * 100:.2f}% de probabilidade de inadimplência.\n")

# 4. Regra de negôcio da Empresa (ponto de corte de 50%0
if probabilidade_risco > 0.5:
   print("CRÉDITO REPROVADO: O modelo indentificou alto risco na operação")
else:
    print("CRÉDITO APROVADO: O perfil encontra-se dentro da margem de segurança.")
```
