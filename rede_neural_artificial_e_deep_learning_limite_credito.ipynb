```python
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Carregar os Dados
df = pd.read_csv('limites_de_credito.csv')
#display(df)
X = df[['Idade', 'Salario', 'Score_Credito']]
y = df['Limite_Aprovado']

# 2. Normalizar X (Entradas ) e também y (Alvo)
normalizador_x = StandardScaler()
X_norm = normalizador_x.fit_transform(X)

normalizador_y = StandardScaler()
y_norm = normalizador_y.fit_transform(y.values.reshape(-1, 1))

# Passar as variáveis normalizadas para treino
X_treino, X_teste, y_treino, y_teste = train_test_split(X_norm, y_norm, test_size=0.2, random_state=42) # ---- Um jeito de começar os neurônios e pegar as entrega das e multiplicar ou quadriplicando

# Construindo a IA
modelo_limite = keras.Sequential([
    keras.Input(shape=(3,)),
    keras.layers.Dense(16, activation='relu'),
    keras.layers.Dense(8, activation='relu'),
    keras.layers.Dense(1)
])

modelo_limite.compile(optimizer='adam', loss='mse', metrics=['mae'])

# Treinar o modelo
print("Treinar o avaliador de limites ...")
# Vamos aumentar para 100 épocas para ter melhor precisão
modelo_limite.fit(X_treino, y_treino, epochs=70)
print("Treino Concluído!\n")

# 5. Testando com, o nosso cliente real
# Cliente tem 35 anos, ganha $ 8.000 e tem Score de 750
cliente_novo = pd.DataFrame({'Idade': [35], 'Salario': [8000], 'Score_Credito': [750]})
client_norm = normalizador_x.transform(cliente_novo)

# A IA vai prever um número pequeno (ex: 0.15)
limite_previsto_norm = modelo_limite.predict(client_norm)

# Usamos o normalizador para reverte o número para Reais
limite_real = normalizador_y.inverse_transform(limite_previsto_norm)

print(f"O limite de Crédito sugerido pela IA é: R$ {limite_real[0][0]:.2f}")
```
