```python
# Preparando o terrreno
import pandas as pd
from sklearn.linear_model import LinearRegression

# Criando os nosso dados históricos (o passado para encinar a máquina)
dados = {
    'Tamanho_m2': [50, 70, 80, 100, 120],
    'Valor': [150000, 290000, 240000, 300000, 360000]
}

df = pd.DataFrame(dados)
print("Os nossos dados de trino: ")
display(df)
```

```python
# Gerar um gráfico de dispersão (Scatter Plot) ---> Metodo dos minimos quadrados
df.plot(kind='scatter', x='Tamanho_m2', y='Valor', color='blue', title='Relação: Tamanho vs valor')
```

```python
# Separando X e Y
x = df[['Tamanho_m2']]
y = df['Valor']
```

```python
# Criando e Treinando a IA
modelo_ia = LinearRegression() # Criamos a "cérebor" vazio

# Vamos trinar o modelo dando-lhe os Tamanhos (X) e o valores (Y)
modelo_ia.fit(x, y)

print("A Inteligência Artificial foi treinada com sucesso!")
```

```python
# A Hora da Verdade (Predição)

# Criamos uma mini-tabela com 1 linha para a nova casa, com o mesmo nome da coluna
tamanho_nova_casa = pd.DataFrame({'Tamanho_m2': [90]}) # Queremos saber o valor de uma casa de 90m2

# Usamos o comando .predict() para a máquina da a predição
valor_previsto = modelo_ia.predict(tamanho_nova_casa)

print(f"A IA avaliou a casa em: {valor_previsto[0]:.2f} Reais")
```
