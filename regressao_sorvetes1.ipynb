```python
# Dados e o Gráfico de Dispersão
import pandas as pd

# Dados históricos de vendas da loja
dados_sorvetes= {
    'temperatura_c':[20, 22, 25, 28, 30, 35],
    'sorvetes_vendidos': [50, 65, 80, 100, 120, 180]
}

df_sorvetes = pd.DataFrame(dados_sorvetes)
print("o nosso histórico de vendas:")
display(df_sorvetes)
```

```python
# Vamos visualizar a correlação ----> Acuracia
df_sorvetes.plot(kind='scatter', x='temperatura_c', y='sorvetes_vendidos', color='orange', title='Temperatura vs Vendas de Sorvetes')
```

```python
# Treino e a Previsão de IA
from sklearn.linear_model import LinearRegression

# 1. Separar a Pergunta (X) da Resposta (y)
X = df_sorvetes[['temperatura_c']]
y = df_sorvetes['sorvetes_vendidos']

# 2. Treinar o Cérebro da IA
modelo_sorvetes = LinearRegression()
modelo_sorvetes.fit(X, y)

# 3. A Predição: Amanhã vai fazer uma onde de calor de 38 graus!
temperatura_amanha = pd.DataFrame({'temperatura_c': [38]})
predicao = modelo_sorvetes.predict(temperatura_amanha)

print(f"ALERTA DE ESTOQUE: A IA prevê que amanhã vai vender {int(predicao[0])} sorvetes!")
```
